```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv("q4-data.csv")

# Extract only features (ignore label column)
X = df[['Feature1', 'Feature2']]

# Apply KMeans clustering with k=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X)

# Predict cluster labels
cluster_labels = kmeans.labels_

# Calculate Silhouette Score
silhouette = silhouette_score(X, cluster_labels)

print(f"Silhouette Score for k=3: {silhouette:.4f}")

# Optional: Visualize the clusters
plt.scatter(X['Feature1'], X['Feature2'], c=cluster_labels, cmap='viridis', s=50)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='x', s=200, label='Centroids')
plt.title("K-Means Clustering (k=3)")
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Load the dataset
df = pd.read_csv("q4-data.csv")
X = df[['Feature1', 'Feature2']]

# Elbow method
sse = []
k_values = list(range(2, 9))

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    sse.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(8, 5))
plt.plot(k_values, sse, 'bo-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Sum of Squared Errors (SSE)')
plt.title('Elbow Method for Optimal k')
plt.grid(True)
plt.show()

# Output SSE for each k
for k, s in zip(k_values, sse):
    print(f"k = {k}, SSE = {s:.2f}")
```

```python
import pandas as pd
from scipy.cluster.hierarchy import linkage, cophenet
from scipy.spatial.distance import pdist

# Load data
data = pd.read_csv('q4-data.csv')
X = data[['Feature1', 'Feature2']]

# Apply Hierarchical Agglomerative Clustering with Single Linkage
Z = linkage(X, method='single')

# Calculate Cophenetic Correlation Coefficient
c, coph_dists = cophenet(Z, pdist(X))

print("Cophenetic Correlation Coefficient:", round(c, 4))
```
